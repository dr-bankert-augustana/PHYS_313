<font size=5><b>3.1 Creating DataFrames From Dictionaries</b></font>

<p align="justify">
One of the most straightforward ways to create a DataFrame in Pandas is by using a Python dictionary. In this approach, each key in the dictionary represents a column name, and the corresponding value is a list (or array) containing the data for that column. Pandas automatically aligns these values by row, as long as the lists are of equal length.
</p>

<br>

<p align="justify">
This method is especially useful when you already have structured data in Python (for example, experimental measurements or simulation results stored in variables) and you want to organize them into a tabular format for further analysis.
</p>

<br>

<b>Code Example:</b>

In the following code cell, we create a simple DataFrame from some synthetic experimental data.

```python
# Import the Pandas Library:

import pandas as pd

# Create a dictionary of data:

data = {
    "Time (s)": [0, 1, 2, 3, 4],
    "Position (m)": [0, 2.5, 10, 22.5, 40],
    "Velocity (m/s)": [0, 5, 10, 15, 20]
}

# Convert the dictionary into a DataFrame:

df = pd.DataFrame(data)

# Display the DataFrame:

display(df)
```

<b>Understanding the Output</b>

<p align="justify">
This printed table looks very similar to what you'd see in a spreadsheet like Excel or Google Sheets:
</p>

  * <b>Columns:</b> Each column corresponds to one of the keys from the dictionary. In this case, we have three columns: <code>Time (s)</code>, <code>Position (m)</code>, and <code>Velocity (m/s)</code>. These serve the same purpose as column headers in a spreadsheet, labeling the type of data in each column.

  * <b>Rows:</b> Each row contains a single set of related data points; one time value, one position, and one velocity. These rows represent observations or measurements taken at different points in time.

  * <b>Index column (on the far left):</b> By default, Pandas automatically adds an index column starting from 0. This index uniquely identifies each row, similar to row numbers in a spreadsheet. While it doesn't carry physical meaning in this example, it becomes very useful when accessing or referencing specific rows programmatically.

<br>

<p align="justify">
So, in essence, the Pandas DataFrame behaves much like a digital spreadsheet, but with the added power of Python. As we see in this lesson, you can apply mathematical operations, filter rows, compute statistics, and generate plots all from this structured format, making it a key tool in your computational physics toolkit.
</p>

<font size=5><b>5.1 Bracket Notation []</b></font>

<p align="justify">
Once we have loaded and explored our DataFrame, the next step is accessing specific columns or subsets of data. One of the most common and flexible ways to do this is using bracket selection, which looks like: <code>DataFrame[selector]</code>
</p>

<br>

<p align="justify">
Depending on what we put inside the brackets, we can select a column, multiple columns, or filter rows using logical conditions.
</p>

<center>$\rule$</center>
<font size=5><b>5.2 Selecting a Single Column</b></font>

<p align="justify">
If we pass a string (column name) inside the brackets, Pandas returns a Series (a one-dimensional view of that column). This works like selecting a column in a spreadsheet by its header. The result behaves like a list but with labels (index) attached.
</p>

<br>

><b>Note:</b><p align="justify">
We can also call up a column by using: <code>DF_Name.Column_Name</code>
</p>

<br>

<b>Code Example:</b>

```python
# Select the 'Pressure (Mpa)' column of the full data:

pressures = full_data_df['Pressure (MPa)']

# Display the firstnames data:

display(pressures)
```

<center>$\rule$</center>
<font size=5><b>5.3 Selecting a Multiple Columns</b></font>

<p align="justify">
To select more than one column, pass in a list of column names inside the brackets: <code>subset_df = df[[column_1, column_2]]</code>. Now you get a new DataFrame with just those columns. Similar to hiding other columns in a spreadsheet.
</p>

<br>

<b>Code Example:</b>

```python
# Select the 'Temperature (C)' and 'Pressure (MPa)' columns of the full data DataFrame:

data_subset_df = full_data_df[['Temperature (C)', 'Pressure (MPa)']]

# Display the subset data:

display(data_subset_df)
```

<center>$\rule$</center>
<font size=5><b>5.4 Selecting Rows with Logical Conditions</b></font>

<p align="justify">
You can also use bracket selection with boolean expressions to filter rows. This is incredibly powerful for working with physical measurements or simulation outputs. This is called using a <b>Boolean Mask</b>. Example: <code>subset = df[boolean_mask]</code>. This filters the DataFrame and only returns rows that meet the condition, just like using filters in Excel or Google Sheets.
</p>

><b>Note:</b><p align="justify">
You can build more complex filters by combining multiple conditions.
</p>

<br>

<b>Code Example:</b>

```python
# Create a subset of data with temperature above 100 C:

high_temp = full_data_df[full_data_df['Temperature (C)'] > 100]

# Display the subset data:

print("High Temperature Data:")
display(high_temp.head())
print()

# Create a subset of data with temperatures between 50 C and 100 C:

medium_temp = full_data_df[(full_data_df['Temperature (C)'] > 50) & (full_data_df['Temperature (C)'] < 100)]

# Display the subset data:

print("Medium Temperature Data:")
display(medium_temp.head())
print()
display(medium_temp.tail())
print()

# Create a subset of data with temperatures below 50 C or above 150 C:

high_and_low_temp = full_data_df[(full_data_df['Temperature (C)'] < 50) | (full_data_df['Temperature (C)'] > 150)]

# Display the subset data:

print("High and Low Temps Only:")
display(high_and_low_temp.head())
print()
display(high_and_low_temp.tail())
```

<center>$\rule$</center>
<font size=5><b>5.5 <code>.loc()</code> and <code>.iloc()</code></b></font>

<p align="justify">
Pandas provides two powerful accessors:
</p>

  * .loc[]: selects by label

  * .iloc[]: selects by integer position

<br>

<p align="justify">
These are much more versatile than [] alone, especially when selecting specific rows and columns at the same time or when doing row slicing.
</p>

  * .loc[]: Selection by Label

    * Syntax: <code>df.loc[row_selector, column_selector]</code>

  * Selecting Rows by Index Label

    * Syntax: <code>df.loc[0]  # Selects the row with index label 0</code>

  * Row Slicing by Label

    * Syntax: <code>df.loc[0:2]</code>

    ><b>Note:</b> <code>.loc[]</code> slicing includes the stop value (unlike standard Python slicing).

  * Selecting Specific Rows and Columns by Label

    * Syntax: <code>df.loc[0:2, ["Time (s)", "Velocity (m/s)"]]</code>

    * This returns rows 0 through 2, and only the "Time" and "Velocity" columns.

  * .iloc[]: Selection by Integer Position

    * Syntax: df.iloc[row_index, column_index]

    * Use .iloc[] when you want to slice or select based on numerical index positions, just like Python lists.

  * Selecting a Row by Integer Index

    * df.iloc[0]  # First row

  * Row Slicing by Position

    * df.iloc[0:3]  # Rows 0 through 2 (excludes stop index, like Python slicing)

  * Selecting Specific Rows and Columns by Position

    * df.iloc[0:3, 0:2]  # First 3 rows, first 2 columns

```python
# Use .loc[] to select specific rows and columns by their labels:

selected_data = full_data_df.loc[10:20, ['Temperature (C)', 'Pressure (MPa)']]

# Display the selected data:

print("Selected Data Using .loc[]:")
display(selected_data)
print()
```

```python
# Use .iloc[] to select rows and columns by their integer positions:

selected_data = full_data_df.iloc[10:20, 0:2]

# Display the selected data:

print("Selected Data Using .iloc[]:")
display(selected_data)
print()
```

<center>$\rule$</center> <font size=5><b>5.6 Using the <code>query()</code> Function</b></font> <p align="justify"> We have seen that we can use Boolean masks to carve out specific subsets of data from our DataFrames. However, the syntax can sometimes look a little clunky, especially when we have multiple conditions. </p> <p align="justify"> Fortunately, Pandas provides a simpler, more natural-feeling way to accomplish the same thing using the <b><code>query()</code></b> function. </p>

For example, suppose we want to find all of the saturated-water data where the temperature is above 100°C:

```python
high_temp = full_data_df.query('`Temperature (C)` > 100')

display(high_temp.head())
```

<b>Multiple Conditions</b>

We can also combine conditions using 'and' and 'or'.

For example, let's find data where the temperature is between 50°C and 100°C:

```python
medium_temp = full_data_df.query(
    '`Temperature (C)` >= 50 and `Temperature (C)` <= 100'
)

display(medium_temp.head())
```

<p align="justify"> Let's suppose that you aren't interested in all of the data in our saturated-water DataFrame, but are instead interested in a specific list of temperatures. </p> <p align="justify"> For example, suppose we want to examine the properties of saturated water at a few temperatures that are particularly relevant to our thermodynamics applications. We'll start by making a list of those temperatures. </p>

```python
list_of_temperatures = [0, 25, 50, 75, 100]
```

Next, we'll look at our DataFrame and pick out only the rows whose temperature is in that list using the <code>isin()</code> function.

```python
selected_temperatures = full_data_df[
    full_data_df["Temperature (C)"].isin(list_of_temperatures)
]

display(selected_temperatures)
```

<p align="justify"> Again, we can accomplish this task using the <code>query()</code> function. This time, instead of using <code>isin()</code>, we can use the <b><code>in</code></b> operator within our query. </p>

For example, let's use our list of temperatures from above:

```python
selected_temperatures2 = full_data_df.query(
    "`Temperature (C)` in @list_of_temperatures"
)

display(selected_temperatures2)
```

<p align="justify"> As before, the <b><code>query()</code></b> function makes it easy and natural to ask multiple questions of our data at once. </p>

<p align="justify"> For example, let's start with our list of selected temperatures, but now we'll add another condition. Suppose we only want the temperatures in our list that correspond to sound speeds <b>greater than or equal to 450 m/s</b>. </p>

```python
selected_data = full_data_df.query(
    "`Temperature (C)` in @list_of_temperatures and `Sound Spd. (v, m/s)` >= 450"
)

display(selected_data)
```

## Group Saturated Water Data by Temperature Range

Let's say that we want to find out information about specific temperature ranges within our DataFrame. We can do so by grouping our entries based on some criteria.

For example, let's group the saturated-water data based on the temperature of the water.

To make our groups more useful, we will create three temperature categories:

1) Low Temperature: below 50°C
2) Medium Temperature: 50°C - 100°C
3) High Temperature: above 100°C

We will then examine the temperature, saturation pressure, and enthalpy of the water within each group.

```python
# Create a new column that categorizes the data based on temperature:

full_data_df["Temperature Range"] = pd.cut(
    full_data_df["Temperature (C)"],
    bins=[-float("inf"), 50, 100, float("inf")],
    labels=["Low Temperature", "Medium Temperature", "High Temperature"]
)

# Group the data by temperature range:

grouped_by_temp = full_data_df.groupby(
    "Temperature Range",
    observed=True
)[
    ["Temperature (C)", "Pressure (MPa)", "Enthalpy (l, kJ/kg)"]
]
```

Let's now examine our new grouping.

```python
print("Grouped By Temperature Range DataFrame Head:")

print()

print(grouped_by_temp.head())

print()
print()

print("Grouped By Temperature Range DataFrame Tail:")

print()

print(grouped_by_temp.tail())
```

It's hard to tell what's going on just from the above. Python has created a group for each temperature range and assigned each row to its appropriate group.

We can, however, perform operations on the groupings.

For example, we can find the average temperature, saturation pressure, and enthalpy of the water within each temperature range using the mean() function.

<b>
numeric_only = True
</b>

This tells Python that we only want to calculate averages for columns that contain numerical values. This is particularly useful when our DataFrame also contains categorical or text-based columns, such as our "Temperature Range" column.

Let's calculate the average properties for each temperature range:

```python
average_properties_by_range = full_data_df.groupby(
    "Temperature Range",
    observed = True
).mean(numeric_only=True)

average_properties_by_range
```

We can instead find the least value in a group by using the <b>min()</b> function

For example, let's find the minimum enthalpy within each of our temperature ranges.

```python
least_enthalpy = minimum_properties_by_range = full_data_df.groupby(
    "Temperature Range",
    observed = True
)["Enthalpy (l, kJ/kg)"].min()

display(least_enthalpy)
```

Or, we can find the largest value within each group by using the <b>max()</b> function.

```python
most_enthalpy = minimum_properties_by_range = full_data_df.groupby(
    "Temperature Range",
    observed = True
)["Enthalpy (l, kJ/kg)"].max()

display(most_enthalpy)
```

We can even make up our own functions to parse the data by using the <b>agg()</b> function which stands for "aggregate".

For example, let's create a function that calculates the range of a thermodynamic property within each temperature range.

```python
def property_range(value_list):

    largest = max(value_list)

    smallest = min(value_list)

    return largest - smallest

pressure_range_by_temp = full_data_df.groupby(
    "Temperature Range",
    observed=True
)[["Pressure (MPa)"]].agg(property_range)

pressure_range_by_temp = pressure_range_by_temp.rename(
    columns={"Pressure (MPa)": "Pressure Range (MPa)"}
)

display(pressure_range_by_temp)
```

<b>Some useful built-in functions</b>

| Function | What it calculates | Example          |

| `max`    | Largest value      | `.agg("max")`    |
| `sum`    | Sum of all values  | `.agg("sum")`    |
| `count`  | Number of values   | `.agg("count")`  |
| `median` | Middle value       | `.agg("median")` |
| `std`    | Standard deviation | `.agg("std")`    |
